## Flow to pressure (air)

In the context of syringe pumps, pressure is closely related to flow rate.



To find the air pressure generated by a syringe pump for a given flow rate, use the Hagen-Poiseuille equation adapted for pressure drop: $$\Delta P = \frac{8 \mu L Q}{\pi r^4}$$. 

Here, $\Delta P$ is the pressure difference, $\mu$ is air viscosity ($1.81 \times 10^{-5}\text{ Pa}\cdot\text{s}$), $L$ and $r$ are the length and radius of the tubing/needle, and $Q$ is the volumetric flow rate.

### Equation and Variables

* **Formula:** $\Delta P = \frac{8 \mu L Q}{\pi r^4}$
* $\Delta P$ = Pressure drop (Pascals, $\text{Pa}$)
* $\mu$ = Dynamic viscosity of air $\approx 1.81 \times 10^{-5}\text{ Pa}\cdot\text{s}$ (at room temperature)
* $L$ = Length of the outlet tube, needle, or restriction (meters, $\text{m}$)
* $Q$ = Volumetric flow rate ($\text{m}^3/\text{s}$)
* $r$ = Internal radius of the tubing or needle (meters, $\text{m}$)
* Total upstream pressure: $P_{in} = P_{atm} + \Delta P$ (where $P_{atm}$ is atmospheric pressure)

### Important Nuances for Air Systems

* **Compressibility:** Air is a compressible gas. This equation works well for low flow rates and small pressure drops ($\Delta P \ll P_{atm}$). If pressures are high, gas expansion and compressibility effects change the linearity.
* **Laminar Flow Check:** This formula assumes laminar flow. Calculate the Reynolds number ($Re = \frac{4\rho Q}{\pi \mu d}$, with $\rho$ as air density $\approx 1.2\text{ kg}/\text{m}^3$). It must remain below 2,100.
* **Syringe Compliance:** Air acts like a spring inside the syringe barrel. At the start of pumping, the pump motor must compress the trapped air volume before steady-state flow and target pressure are achieved downstream.
 

```python
import math

def calculate_pressure(flowrate, length, diameter, viscosity=18.5e-6):
    """
    Calculate the pressure drop in a pipe using the Hagen-Poiseuille equation.

    Parameters:
    flowrate (float): The volumetric flow rate (µl/min).
    length (float): The length of the pipe (cm).
    diameter (float): The diameter of the pipe (mm).
    viscosity (float): The dynamic viscosity of the fluid (Pa·s).

    Returns:
    float: The pressure drop (Pa).
    """    

    # Convert 
    flowrate_m3_s = flowrate * 1e-9 / 60  # µl/min to m^3/s
    length_m = length / 100  # cm to m
    diameter_m = diameter / 1000  # mm to m

    # Calculate the pressure drop using the Hagen-Poiseuille equation
    pressure_drop = (8 * viscosity * length_m * flowrate_m3_s) / (math.pi * (diameter_m/2)**4)
    #pressure_drop = (1.23e-5 * length * flowrate) / (math.pi * (diameter/2)**4)

    return pressure_drop/100

def calculate_reynolds(flowrate, diameter, density=1000, viscosity=18.5e-6):
    """
    Calculate the Reynolds number for flow in a pipe.

    Parameters:
    flowrate (float): The volumetric flow rate (µl/min).
    diameter (float): The diameter of the pipe (mm).
    density (float): The density of the fluid (kg/m^3).
    viscosity (float): The dynamic viscosity of the fluid (Pa·s).

    Returns:
    float: The Reynolds number.
    """

    # Convert 
    flowrate_m3_s = flowrate * 1e-9 / 60  # µl/min to m^3/s
    diameter_m = diameter / 1000  # mm to m

    # Calculate the average velocity
    area = math.pi * (diameter_m / 2) ** 2
    velocity = flowrate_m3_s / area

    # Calculate Reynolds number
    reynolds_number = (density * velocity * diameter_m) / viscosity

    return reynolds_number
```

```python
print(f"Pressure drop: {calculate_pressure(flowrate=500, length=1000, diameter=0.4)} mbar")

print(f"Reynolds number: {calculate_reynolds(flowrate=500, diameter=0.5)}")
```
